```python
import os
os.makedirs("../images", exist_ok=True)

palette = ["#4C78A8", "#F58518", "#54A24B", "#E45756", "#72B7B2",
           "#B279A2", "#FF9DA6", "#9D755D", "#BAB0AC", "#EECA3B"]

def make_pie(data, title, filename):
    data.plot(kind="pie", autopct="%1.1f%%", startangle=90, figsize=(6, 6),
              colors=palette[:len(data)], ylabel="")
    plt.title(title)
    plt.tight_layout()
    plt.savefig("../images/" + filename, dpi=150)
    plt.show()

for col in ["Category", "Payment Mode", "Region", "Segment"]:
    data = df.groupby(col)["Sales"].sum().sort_values(ascending=False)
    make_pie(data, "Sales share by " + col, "pie_sales_by_" + col.lower().replace(" ", "_") + ".png")

state_sales = df.groupby("State")["Sales"].sum().sort_values(ascending=False)
top5 = state_sales.head(5)
data = pd.concat([top5, pd.Series({"Others": state_sales.iloc[5:].sum()})])
make_pie(data, "Sales share: Top 5 States + Others", "pie_sales_by_state.png")

print("Images saved:", os.listdir("../images"))
```
